```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
kb_name = ["family", "mutagenesis", "vicodi", "carcinogenesis"]

for dataset in kb_name:
    df = pd.read_csv(f"cache_memory_per_cel_{dataset}.csv")

    # ---------------------------------------------------------------------------
    # Grouped bar plot — 2 bars per CEL: Semantic Cache vs Non-Semantic Cache
    # ---------------------------------------------------------------------------
    CEL_ORDER = ["OCEL", "CELOE", "EvoLearner", "CLIP"]
    CACHE_MODES = ["semantic", "non_semantic"]

    # Colours matching the reference screenshot (blue solid / green hatched)
    MODE_STYLE = {
        "semantic":     dict(color="dodgerblue", hatch="",    label="Semantic Cache"),
        "non_semantic": dict(edgecolor="green", hatch="///", label="Non-Semantic Cache", facecolor="none", linewidth=2),
    }

    width = 0.25
    x     = np.arange(len(CEL_ORDER))

    # Extract values in CEL_ORDER order explicitly
    semantic_vals     = [df[(df["mode"] == "semantic")     & (df["CEL"] == cel)]["cache_memory_MB"].values[0] for cel in CEL_ORDER]
    non_semantic_vals = [df[(df["mode"] == "non_semantic") & (df["CEL"] == cel)]["cache_memory_MB"].values[0] for cel in CEL_ORDER]

    fig, ax = plt.subplots(figsize=(9, 9))

    ax.bar(x-width, semantic_vals,     width=width, **MODE_STYLE["semantic"])
    ax.bar(x,       non_semantic_vals, width=width, **MODE_STYLE["non_semantic"])

    ax.set_xticks(x)
    ax.set_xticklabels(CEL_ORDER, fontsize=12)
    ax.set_xlabel("", fontsize=12)
    ax.set_ylabel("Average Memory (MB)", fontsize=20)

    ax.set_xticks(x)
    ax.tick_params(axis='y', labelsize=20)
    ax.tick_params(axis='x', labelsize=20)

    # Grid
    ax.grid(True, axis="y", linestyle="--", linewidth=0.6, color="lightgray")

    # Legend with larger font
    ax.legend(fontsize=20)

    plt.xticks(fontsize=20)
    plt.yticks(fontsize=20)

    # Save as PDF
    filename = f"memory_results_{dataset}.pdf"
    plt.savefig(filename, bbox_inches='tight')
    plt.close()

    print(f"Saved: {filename}")
```
